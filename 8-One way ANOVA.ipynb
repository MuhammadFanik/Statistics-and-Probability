```python
from scipy.stats import f_oneway
```

```python
campaign_a = [32, 36, 38, 40, 43, 45]
campaign_b = [44, 47, 49, 51, 53, 56]
campaign_c = [55, 58, 60, 62, 65, 66]
```

```python
f_stat, p_value = f_oneway(
    campaign_a,
    campaign_b,
    campaign_c
)
```

```python
print("F-statistic:", f_stat)
print("P-value:", p_value)
```
