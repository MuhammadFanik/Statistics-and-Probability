```python
import numpy as np
```

```python
old_design = np.array([120, 130, 110, 115, 125])
new_design = np.array([100, 110, 90, 95, 105])
```

```python
mean_old = np.mean(old_design)
mean_new = np.mean(new_design)
```

```python
std_old = np.std(old_design, ddof=1)
std_new = np.std(new_design, ddof=1)
```

```python
# Pooled standard deviation
n1 = len(old_design)
n2 = len(new_design)
```

```python
pooled_std = np.sqrt(
    ((n1 - 1) * std_old**2 + (n2 - 1) * std_new**2)
    / (n1 + n2 - 2))
```

```python
cohens_d = (mean_old - mean_new) / pooled_std
```

```python
print("Cohen's d:", cohens_d)
```
