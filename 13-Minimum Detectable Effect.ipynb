```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize
```

```python
baseline_rate = 0.10
```

```python
analysis = NormalIndPower()
```

```python
# Calculate the detectable standardized effect size
effect_size = analysis.solve_power(
    nobs1=1000,
    alpha=0.05,
    power=0.80,
    ratio=1.0,
    alternative="two-sided"
)
```

```python
# Convert standardized effect size into a conversion rate
# using Cohen's arcsine effect-size convention.
# Here we solve for an increase from the baseline rate.

import numpy as np

mde_rate = np.sin(
    2 * (np.arcsin(np.sqrt(baseline_rate)) + effect_size / 2)
) ** 2
```

```python
print("Baseline conversion:", baseline_rate)
print("Detectable conversion:", mde_rate)
print("Absolute MDE:", mde_rate - baseline_rate)
```
