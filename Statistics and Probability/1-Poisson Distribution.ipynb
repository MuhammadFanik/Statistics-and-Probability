# Poisson Distrubtion

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
# Average number of events (lambda)
lambda_val = 4      # e.g: 4 calls per hour
# Number of events we want probabilities for
k_values = np.arange(0, 11)

# PMF (X=k)
pmf_vals = poisson.pmf(k_values, mu=lambda_val)

# CDF (X<=k)
cdf_vals = poisson.cdf(k_values, mu=lambda_val)

# Print probabilities
for k, pmf, cdf in zip(k_values, pmf_vals, cdf_vals):
    print(f"Events = {k}, PMF = {pmf:.4f}, CDF = {cdf:.4f}")
```

```python
# Plot PMF
plt.figure(figsize=(10,5))
plt.bar(k_values, pmf_vals, color='skyblue', alpha=0.7)
plt.plot(k_values, cdf_vals, 'o-', color='red', label='CDF')
plt.xlabel('Number of events (k)')
plt.ylabel('Probability')
plt.title(f'Poisson Distribution (lambda={lambda_val})')
plt.legend()
plt.show()
```
