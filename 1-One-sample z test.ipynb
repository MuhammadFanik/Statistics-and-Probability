```python
import numpy as np
from scipy.stats import norm
```

```python
# Given data
pop_mean = 50
pop_std = 10
sample_mean = 53
sample_size = 100
```

```python
# Calculate the z-statistic
z_statistic = (sample_mean - pop_mean) / (pop_std / (sample_size ** 0.5))
print(z_statistic)
```

```python
# Calculate two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z_statistic)))
```

```python
# Significance level
alpha = 0.05
```

```python
print(f"Z-statistic: {z_statistic}")
print(f"p-value: {p_value}")
```

```python
if p_value < z_statistic:
    print("We have enough evidence to reject the null hypothesis.")
else:
    print("We dont have sufficient evidence to reject the null hypothesis.")
```
