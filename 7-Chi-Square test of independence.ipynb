```python
import numpy as np
from scipy.stats import chi2_contingency
```

```python
observed = np.array([
    [60, 40],
    [20, 80]
])
```

```python
chi_stat, p_value, dof, expected = chi2_contingency(
    observed,
    correction=False
)
```

```python
print("Chi-Square:", chi_stat)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:\n", expected)
```
