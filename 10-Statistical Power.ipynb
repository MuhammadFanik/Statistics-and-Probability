```python
from statsmodels.stats.power import TTestIndPower
```

```python
analysis = TTestIndPower()
```

```python
power = analysis.power(
    effect_size=0.5,    # Assumed standardized difference between group means
    nobs1=64,       # Sample Size in the first group
    alpha=0.05,     # Significance level
    ratio=1.0,      # Both groups have equal sample size
    alternative="two-sided"     # We test for a difference in either direction
)
```

```python
print("Statistical Power:", power)
```
