```python
import pandas as pd
```

```python
data = {
    "campaign": (
        ["A"] * 8 +
        ["B"] * 8 +
        ["C"] * 8),

    "membership": (
        ["Regular"] * 4 + ["Premium"] * 4) * 3,

    "spending": [
        30, 32, 31, 33,
        50, 52, 51, 53,

        40, 42, 41, 43,
        60, 62, 61, 63,

        50, 52, 51, 53,
        70, 72, 71, 73]}

df = pd.DataFrame(data)
```

```python
df
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
model = ols(
    "spending ~ C(campaign) * C(membership)",
    data=df
).fit()
```

```python
# - spending: Dependent variable.
# - C(campaign): First categorical factor.
# - C(membership): Second categorical factor.
# - *: Includes both main effects and their interaction.
# This is important: using + instead of * would include only the main effects, not their interaction.
```

```python
anova_table = sm.stats.anova_lm(
    model,
    typ=2)
```

```python
print(anova_table)
```

```python
# Effect	                         Expected result
# Campaign	                         Significant
# Membership	                     Significant
# Campaign × Membership	             Not significant
#
# Why?
# - Campaign means are 31, 41, and 51 for regular customers, and 51, 61, and 71 for premium customers. Campaign differences are consistent.
# - Premium customers spend 20 more than regular customers in every campaign.
# - The difference between premium and regular customers does not change across campaigns.
```
