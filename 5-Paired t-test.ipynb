```python
import numpy as np
from scipy.stats import ttest_rel
```

```python
# Customer spending before the system
before = np.array([40, 55, 35, 60, 50])
```

```python
# Customer spending after the system
after = np.array([50, 60, 42, 68, 55])
```

```python
# Calculate individual differences
differences = after - before
```

```python
# Perform paired t-test
t_stat, p_value = ttest_rel(
    after,
    before,
    alternative="two-sided"
)
```

```python
print("Mean difference:", np.mean(differences))
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```
