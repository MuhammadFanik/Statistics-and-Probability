```python
import numpy as np
from scipy.stats import ttest_ind
```

```python
# Campaign A customer spending
campaign_a = np.array([
    45, 50, 55, 48, 52, 60, 42, 51, 49, 53,
    47, 56, 50, 44, 58, 52, 46, 54, 49, 57
])
```

```python
# Campaign B customer spending
campaign_b = np.array([
    55, 60, 58, 62, 54, 65, 59, 57, 63, 61,
    56, 68, 60, 64, 53, 66, 58, 62, 59, 67
])
```

```python
# Welch's Independent two-sample t-test
t_stat, p_value = ttest_ind(
    campaign_a,
    campaign_b,
    equal_var=False,
    alternative="two-sided"
)
```

```python
print("Campaign A Mean:", np.mean(campaign_a))
print("Campaign B Mean:", np.mean(campaign_b))
print("T-statistic:", t_stat)
print("P-value:", p_value)
```
